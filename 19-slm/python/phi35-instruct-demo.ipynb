```python
import torch 
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline 
```

```python
torch.random.manual_seed(0) 
model = AutoModelForCausalLM.from_pretrained( 
    "microsoft/Phi-3-mini-4k-instruct",
    attn_implementation='eager',  
    device_map="cuda",  
    torch_dtype="auto",  
    trust_remote_code=True, 
) 
```

```python
tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct") 
```

```python
# messages = [ 
#     {"role": "system", "content": "你是我的人工智能助手，协助我用中文解答问题"}, 
#     {"role": "user", "content": "你能否介绍一下如何提升学习效率吗?"}, 
# ] 

# <|system|>
# You are a helpful assistant.<|end|>
# <|user|>
# Question?<|end|>
# <|assistant|>

messages = "<|system|>\n You are my AI assistant, helping me answer questions in English.\n<|end|><|user|>\n Do you know Markiplier? ? \n<|end|><|assistant|>"
```

```python
pipe = pipeline( 
    "text-generation", 
    model=model, 
    tokenizer=tokenizer, 
) 
```

```python
generation_args = { 
    "max_new_tokens": 1024, 
    "return_full_text": False, 
    "temperature": 0.5, 
    "do_sample": False, 
} 
```

```python
output = pipe(messages, **generation_args) 
```
